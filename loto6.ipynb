```python
import pandas as pd
loto = pd.read_csv('loto6.csv', encoding='shift_jis')
loto.head()
```

```python
lotoc = loto.copy()
```

```python
numbers = lotoc[['第1数字', '第2数字', '第3数字', '第4数字', '第5数字', '第6数字', 'BONUS数字']]
numbers
```

```python
allnum = numbers.values.reshape(-1)
allnum
```

```python
s = pd.Series(allnum)
freq = s.value_counts().sort_index()
freq
```

```python
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = 'Meiryo'
plt.rcParams['axes.unicode_minus'] = False

plt.figure(figsize=(14,4))
freq.plot(kind='bar', rot=0)

plt.title("ロト 番号別 出現回数")     
plt.xlabel("番号")                    
plt.ylabel("出現回数")                

plt.tight_layout()
plt.show()
```

```python
top10 = freq.sort_values(ascending=False).head(10)
top10
```
